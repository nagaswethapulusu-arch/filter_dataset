```python
# ===== Task 19: Filter Dataset Records (Pandas) =====
import pandas as pd
import numpy as np
from IPython.display import display

# ---------- 1. Create and load the dataset ----------
rng = np.random.default_rng(7)
n = 50
hours  = np.clip(rng.normal(5, 2, n), 0.5, 10).round(1)
attend = np.clip(rng.normal(85, 8, n), 55, 100).round(0).astype(int)
prev   = np.clip(rng.normal(65, 12, n), 25, 100).round(0).astype(int)
final  = np.clip(0.35*prev + 4*hours + 0.25*attend + rng.normal(0, 5, n), 0, 100).round(0).astype(int)

df = pd.DataFrame({
    'student_id': np.arange(1, n+1),
    'gender': rng.choice(['Male', 'Female'], n),
    'hours_studied': hours,
    'attendance_pct': attend,
    'previous_score': prev,
    'final_score': final})
df['result'] = np.where(df['final_score'] >= 50, 'Pass', 'Fail')

df.to_csv('student_performance.csv', index=False)
df = pd.read_csv('student_performance.csv')
print('Total records:', len(df))
display(df.head())

def show(title, explanation, result):
    print('\n' + '=' * 70)
    print(title)
    print('Explanation:', explanation)
    print('Records found:', len(result))
    display(result)

# ---------- 2. Filtering examples ----------
high_scorers = df[df['final_score'] > 70]
show('Example 1: Single condition  ->  final_score > 70',
     'Keeps only students who scored above 70.', high_scorers)

failed = df[df['result'] == 'Fail']
show("Example 2: Text condition  ->  result == 'Fail'",
     'Keeps only students whose result is Fail.', failed)

dedicated = df[(df['gender'] == 'Female') & (df['hours_studied'] >= 5) & (df['attendance_pct'] > 85)]
show('Example 3: AND (&)  ->  Female, hours >= 5, attendance > 85',
     'A row is kept only if ALL conditions are true (strict).', dedicated)

either = df[(df['final_score'] > 60) | (df['hours_studied'] > 6)]
show('Example 4: OR (|)  ->  final_score > 60 OR hours_studied > 6',
     'A row is kept if AT LEAST ONE condition is true (broader).', either)

not_passed = df[~(df['result'] == 'Pass')]
show("Example 5: NOT (~)  ->  not (result == 'Pass')",
     'The ~ symbol reverses the condition.', not_passed)

selected = df[df['student_id'].isin([1, 5, 10, 15, 20])]
show('Example 6: isin()  ->  student_id in [1, 5, 10, 15, 20]',
     'Keeps rows whose value matches any value in the list.', selected)

mid_range = df[df['final_score'].between(55, 65)]
show('Example 7: between()  ->  final_score between 55 and 65',
     'Keeps rows whose value lies in the range (both ends included).', mid_range)

q = df.query('final_score > 65 and attendance_pct >= 85')
show('Example 8: query()  ->  final_score > 65 and attendance_pct >= 85',
     'Same as Boolean filtering, written as a string.', q)

show('Example 9: loc  ->  filter rows AND choose columns',
     'Filters final_score > 70 and shows only 3 columns.',
     df.loc[df['final_score'] > 70, ['student_id', 'hours_studied', 'final_score']])

# ---------- 3. Why parentheses matter ----------
print('\n' + '=' * 70)
print('Why parentheses matter')
try:
    df[df['final_score'] > 70 & df['gender'] == 'Male']
except Exception as e:
    print('Without parentheses -> error:', type(e).__name__)
right = df[(df['final_score'] > 70) & (df['gender'] == 'Male')]
print('With parentheses -> records found:', len(right))

# ---------- 4. Number of records after each filter ----------
print('\n' + '=' * 70)
counts = {
    'All records': len(df),
    'final_score > 70': len(high_scorers),
    "result == 'Fail'": len(failed),
    'Female & hours>=5 & attendance>85': len(dedicated),
    'final_score > 60 OR hours > 6': len(either),
    'final_score between 55 and 65': len(mid_range),
}
display(pd.Series(counts, name='records'))

# ---------- 5. Observations ----------
print('OBSERVATIONS')
print('-' * 50)
print(f'1. {len(high_scorers)} of {len(df)} students scored above 70 ({len(high_scorers)/len(df)*100:.0f}%).')
print(f'2. {len(failed)} students failed.')
print(f'3. OR is broader than AND: {len(either)} records matched at least one OR condition, while only {len(dedicated)} matched all three AND conditions.')
```
